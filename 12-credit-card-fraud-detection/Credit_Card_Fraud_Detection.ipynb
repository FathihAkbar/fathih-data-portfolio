```python
feature_columns = [
    col for col in df.columns
    if col != "Class"
]

X = df[feature_columns]
y = df["Class"]

print("Number of features:", X.shape[1])
print("Number of target:", y.shape[0])
```

```python
X.head()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training:", X_train.shape)
print("Testing :", X_test.shape)

print("\nFraud in training:")
print(y_train.value_counts())

print("\nFraud in testing:")
print(y_test.value_counts())
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
from sklearn.linear_model import LogisticRegression

lr_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

lr_model.fit(
    X_train_scaled,
    y_train
)

y_pred_lr = lr_model.predict(
    X_test_scaled
)

y_prob_lr = lr_model.predict_proba(
    X_test_scaled
)[:, 1]
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    classification_report
)

lr_accuracy = accuracy_score(
    y_test,
    y_pred_lr
)

lr_precision = precision_score(
    y_test,
    y_pred_lr,
    zero_division=0
)

lr_recall = recall_score(
    y_test,
    y_pred_lr,
    zero_division=0
)

lr_f1 = f1_score(
    y_test,
    y_pred_lr,
    zero_division=0
)

lr_roc_auc = roc_auc_score(
    y_test,
    y_prob_lr
)

lr_pr_auc = average_precision_score(
    y_test,
    y_prob_lr
)

print(f"Accuracy : {lr_accuracy:.4f}")
print(f"Precision: {lr_precision:.4f}")
print(f"Recall   : {lr_recall:.4f}")
print(f"F1       : {lr_f1:.4f}")
print(f"ROC-AUC  : {lr_roc_auc:.4f}")
print(f"PR-AUC   : {lr_pr_auc:.4f}")
```

```python
print(
    classification_report(
        y_test,
        y_pred_lr,
        zero_division=0
    )
)
```

```python
lr_balanced = LogisticRegression(
    max_iter=1000,
    class_weight="balanced",
    random_state=42
)

lr_balanced.fit(
    X_train_scaled,
    y_train
)
```

```python
from sklearn.metrics import (
    precision_recall_curve,
    auc
)
```

```python
precision_curve, recall_curve, thresholds = (
    precision_recall_curve(
        y_test,
        y_prob_balanced
    )
)
```

```python
plt.figure(figsize=(9, 6))

plt.plot(
    recall_curve,
    precision_curve
)

plt.xlabel("Recall")
plt.ylabel("Precision")
plt.title("Precision-Recall Curve — Balanced Logistic Regression")

plt.grid(alpha=0.2)

plt.show()
```

```python
f1_scores = (
    2 * precision_curve[:-1] * recall_curve[:-1]
    / (
        precision_curve[:-1]
        + recall_curve[:-1]
        + 1e-10
    )
)
```

```python
best_idx = np.argmax(
    f1_scores
)

best_threshold = thresholds[
    best_idx
]

best_precision = precision_curve[
    best_idx
]

best_recall = recall_curve[
    best_idx
]

best_f1 = f1_scores[
    best_idx
]

print(
    f"Best Threshold: {best_threshold:.6f}"
)

print(
    f"Precision: {best_precision:.4f}"
)

print(
    f"Recall: {best_recall:.4f}"
)

print(
    f"F1-Score: {best_f1:.4f}"
)
```

```python
y_pred_threshold = (
    y_prob_balanced >= best_threshold
).astype(int)
```

```python
threshold_accuracy = accuracy_score(
    y_test,
    y_pred_threshold
)

threshold_precision = precision_score(
    y_test,
    y_pred_threshold,
    zero_division=0
)

threshold_recall = recall_score(
    y_test,
    y_pred_threshold,
    zero_division=0
)

threshold_f1 = f1_score(
    y_test,
    y_pred_threshold,
    zero_division=0
)

print(
    f"Accuracy : {threshold_accuracy:.4f}"
)

print(
    f"Precision: {threshold_precision:.4f}"
)

print(
    f"Recall   : {threshold_recall:.4f}"
)

print(
    f"F1-Score : {threshold_f1:.4f}"
)
```

```python
cm_threshold = confusion_matrix(
    y_test,
    y_pred_threshold
)

plt.figure(figsize=(6, 5))

sns.heatmap(
    cm_threshold,
    annot=True,
    fmt="d",
    cmap="Blues"
)

plt.title(
    f"Fraud Detection — Threshold {best_threshold:.3f}"
)

plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.show()
```

```python
X_train_sub, X_val, y_train_sub, y_val = train_test_split(
    X_train,
    y_train,
    test_size=0.25,
    random_state=42,
    stratify=y_train
)

print("Train      :", X_train_sub.shape)
print("Validation :", X_val.shape)
print("Test       :", X_test.shape)

print("\nFraud:")
print("Train      :", y_train_sub.value_counts()[1])
print("Validation :", y_val.value_counts()[1])
print("Test       :", y_test.value_counts()[1])
```

```python
scaler_final = StandardScaler()

X_train_sub_scaled = scaler_final.fit_transform(
    X_train_sub
)

X_val_scaled = scaler_final.transform(
    X_val
)

X_test_scaled = scaler_final.transform(
    X_test
)
```

```python
balanced_model_v2 = LogisticRegression(
    max_iter=1000,
    class_weight="balanced",
    random_state=42
)

balanced_model_v2.fit(
    X_train_sub_scaled,
    y_train_sub
)
```

```python
y_prob_val = balanced_model_v2.predict_proba(
    X_val_scaled
)[:, 1]
```

```python
precision_val, recall_val, thresholds_val = (
    precision_recall_curve(
        y_val,
        y_prob_val
    )
)

f1_val = (
    2 * precision_val[:-1] * recall_val[:-1]
    /
    (
        precision_val[:-1]
        + recall_val[:-1]
        + 1e-10
    )
)

best_idx_val = np.argmax(
    f1_val
)

best_threshold_val = thresholds_val[
    best_idx_val
]

print(
    f"Best Validation Threshold: "
    f"{best_threshold_val:.6f}"
)

print(
    f"Validation Precision: "
    f"{precision_val[best_idx_val]:.4f}"
)

print(
    f"Validation Recall: "
    f"{recall_val[best_idx_val]:.4f}"
)

print(
    f"Validation F1: "
    f"{f1_val[best_idx_val]:.4f}"
)
```

```python
y_prob_test = balanced_model_v2.predict_proba(
    X_test_scaled
)[:, 1]

y_pred_final = (
    y_prob_test >= best_threshold_val
).astype(int)
```

```python
final_precision = precision_score(
    y_test,
    y_pred_final,
    zero_division=0
)

final_recall = recall_score(
    y_test,
    y_pred_final,
    zero_division=0
)

final_f1 = f1_score(
    y_test,
    y_pred_final,
    zero_division=0
)

final_accuracy = accuracy_score(
    y_test,
    y_pred_final
)

final_roc_auc = roc_auc_score(
    y_test,
    y_prob_test
)

final_pr_auc = average_precision_score(
    y_test,
    y_prob_test
)

print(f"Final Threshold: {best_threshold_val:.6f}")
print(f"Accuracy : {final_accuracy:.4f}")
print(f"Precision: {final_precision:.4f}")
print(f"Recall   : {final_recall:.4f}")
print(f"F1       : {final_f1:.4f}")
print(f"ROC-AUC  : {final_roc_auc:.4f}")
print(f"PR-AUC   : {final_pr_auc:.4f}")
```

## Key Findings

### 1. Fraud transactions are extremely rare

After removing duplicate records, the dataset contained 67,172 transactions.

The class distribution was:

- Normal: 99.7484%
- Fraud: 0.2516%

This severe class imbalance makes accuracy alone unsuitable as the primary evaluation metric.

### 2. Precision-Recall metrics are important for fraud detection

Because fraudulent transactions represent only a very small portion of the dataset, the analysis focused on Precision, Recall, F1-Score, ROC-AUC, and especially PR-AUC.

PR-AUC provides a more informative view of model performance under highly imbalanced conditions.

### 3. XGBoost achieved the strongest overall performance

Among the evaluated models, XGBoost achieved the highest PR-AUC and F1-Score under the evaluated setup.

The final XGBoost configuration with a validation-selected threshold of approximately 0.9558 achieved:

- Accuracy: 99.95%
- Precision: 92.86%
- Recall: 79.59%
- F1-Score: 85.71%
- ROC-AUC: 97.99%
- PR-AUC: 87.94%

### 4. Threshold optimization improved the balance between precision and recall

The default classification threshold was not assumed to be optimal for fraud detection.

A threshold was selected using the validation set and then applied to the test set.

This produced a better balance between identifying fraudulent transactions and limiting false alarms.

### 5. V14 was the most influential feature

Feature importance analysis identified `V14` as the strongest predictor, with an importance of approximately 0.415.

Other influential features included `V10`, `V4`, `V12`, `V20`, and `V8`.

Because these variables are anonymized transformed features, their individual business meanings cannot be directly interpreted from the dataset.

### 6. The model provides useful risk prioritization

The risk profiling produced substantially different fraud rates across risk groups.

High Risk transactions had an actual fraud rate of approximately 91.76%, while Low Risk transactions had an actual fraud rate of approximately 0.026%.

This indicates that the model can be used not only for binary classification but also for prioritizing transactions for further investigation.

### 7. False positives remain an operational consideration

The final model produced only 6 false positives on the test set, while 20 fraud transactions were missed.

This represents a trade-off between detecting suspicious transactions and minimizing unnecessary investigation alerts.

## Business Recommendations

### High-Risk Transactions

Transactions classified as High Risk should be prioritized for immediate review.

Possible actions include:

- Temporary transaction verification
- Additional authentication
- Manual investigation
- Real-time fraud alerts
- Temporary payment restrictions when appropriate

### Medium-Risk Transactions

Medium-Risk transactions can receive additional monitoring or secondary verification.

Possible actions include:

- Risk-based authentication
- Additional transaction checks
- Behavioral monitoring
- Customer confirmation

### Low-Risk Transactions

Low-Risk transactions generally require less intervention.

Automated processing can be maintained while continuing passive monitoring for unusual behavior.

### Operational Strategy

The model should be used as a decision-support layer rather than an unconditional replacement for fraud investigation.

A production system could combine:

- Machine learning probability
- Transaction rules
- Customer history
- Device information
- Geographic patterns
- Human investigation

to improve fraud detection and reduce unnecessary transaction blocking.
